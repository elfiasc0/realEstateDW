```python
from faker import Faker
import pandas as pd
import random
from datetime import datetime
from datetime import timedelta

fake = Faker('pl_PL')

number_of_rows_1 = 100

#Apartment table
apartment = []
apartment_addresses = []

city_data = {
    "Warsaw": [("Srodmiescie", "00-001"), ("Mokotow", "02-506"), ("Praga Polnoc", "03-705"), ("Wola", "01-001"), ("Zoliborz", "01-001")],
    "Krakow": [("Stare Miasto", "31-001"), ("Kazimierz", "31-056"), ("Podgorze", "30-001"), ("Nowa Huta", "31-001"), ("Bronowice", "30-005")],
    "Wroclaw": [("Srodmiescie", "50-001"), ("Krzyki", "52-100"), ("Fabryczna", "54-000"), ("Psie Pole", "51-200"), ("Stare Miasto", "50-201")],
    "Gdansk": [("Srodmiescie", "80-001"), ("Wrzeszcz", "80-224"), ("Oliwa", "80-300"), ("Nowy Port", "80-465"), ("Zaspa", "80-462")],
    "Poznan": [("Stare Miasto", "61-001"), ("Jezyce", "60-501"), ("Lazarz", "60-202"), ("Wilda", "61-501"), ("Grunwald", "60-209")]
}

city_district_streets = {
    "Warsaw": {
        "Srodmiescie": ["Marszalkowska", "Chmielna", "Swietokrzyska", "Krakowskie Przedmiescie", "Nowy Swiat"],
        "Mokotow": ["Pulawska", "Woronicza", "Limanowskiego", "Narbutta", "Rakowiecka"],
        "Praga Polnoc": ["Zabkowska", "Klopot", "Inzynierska", "Jagiellonska", "Stalowa"],
        "Wola": ["Mlynarska", "Sowinskiego", "Targowa", "Powisle", "Ludwika Rydygiera"],
        "Zoliborz": ["Batory", "Krasinskiego", "Lechitow", "Bialostocka", "Wilenska"]
    },
    "Krakow": {
        "Stare Miasto": ["Florianska", "Karmelicka", "Dluga", "Pilsudskiego", "Lobzow"],
        "Kazimierz": ["Szeroka", "Jozefa", "Wielopole", "Grodzka", "Plac Nowy"],
        "Podgorze": ["Pieciomorgowa", "Mysliwska", "Krolowej Jadwigi", "Jana Pawla II", "Rzeszowska"],
        "Nowa Huta": ["Wandy", "Centralna", "Lipowa", "Wiosny Ludow", "Aleja Roz"],
        "Bronowice": ["Chorazowka", "Kasprowicza", "Odrzanska", "Podgorki", "Swietokrzyska"]
    },
    "Wroclaw": {
        "Srodmiescie": ["Katedralna", "Olawska", "Swidnicka", "Legnicka", "Grabiszynska"],
        "Krzyki": ["Aleja 1 Maja", "Mokotowska", "Sudecka", "Miejska", "Ziebicka"],
        "Fabryczna": ["Obornicka", "Wroclawska", "Osobowicka", "Hutnicza", "Mickiewicza"],
        "Psie Pole": ["Polan", "Zaodrze", "Trzmielowice", "Biskupin", "Olawka"],
        "Stare Miasto": ["Swietego Mikolaja", "Olawska", "Piekarska", "Rynkowska", "Czarnowiejska"]
    },
    "Gdansk": {
        "Srodmiescie": ["Dlugi Targ", "Wielki Mlyn", "Targ Weglowy", "Swietego Ducha", "Podwale Staromiejskie"],
        "Wrzeszcz": ["Alei Grunwaldzka", "Morska", "Jaskowa Dolina", "Bociania", "Karpacka"],
        "Oliwa": ["Piastowska", "Al. Grunwaldzka", "Ostroda", "Abrahama", "Trojmiasto"],
        "Nowy Port": ["Nowe Wrota", "Gdanska", "Dluga", "Podgorna", "Swinoujscie"],
        "Zaspa": ["Aleja Jana Pawla II", "Gdanska", "Zaspa", "Rynkowska", "Polski Czerwony Krzyz"]
    },
    "Poznan": {
        "Stare Miasto": ["Swiety Marcin", "Polwiejska", "Garbary", "Grunwaldzka", "Laskotelna"],
        "Jezyce": ["Zernickie", "Poludniowa", "Pieciomorgowa", "Kraszewskiego", "Laskowicka"],
        "Lazarz": ["Opolska", "Staroleka", "Kroniki", "Laskocze", "Dabrowa"],
        "Wilda": ["Strzeszynska", "Tory", "Czajkowska", "Wlasnosciowa", "Wiosna"],
        "Grunwald": ["Gorzyslawa", "Blazenkowska", "Sienkiewicza", "Mikolajewska", "Przybyszowska"]
    }
}

for i in range(1, number_of_rows_1 + 1):
  apartment_id = i
  city = random.choice(list(city_data.keys()))
  district, postal_code = random.choice(city_data[city])
  street_name = random.choice(city_district_streets[city][district])
  address = street_name + " " + str(random.randint(1, 100)) + "/" + str(random.randint(1,100))
  size = round(random.uniform(20.0, 150.0), 2)
  floor_number = random.randint(0, 30)
  if size < 40:
      number_of_rooms = random.choice([1, 2])
  elif size < 80:
      number_of_rooms = random.choice([2, 3])
  else:
      number_of_rooms = random.choice([3, 4, 5])
  start_date = datetime.strptime("1950-01-01", "%Y-%m-%d")
  end_date = datetime.strptime("2025-01-01", "%Y-%m-%d")
  year = fake.date_between(start_date=start_date, end_date=end_date)

  apartment.append([apartment_id, address, district, city, postal_code, size, floor_number, number_of_rooms, year])
  apartment_addresses.append(address)

#assigning city to each apartment
apartment_city_map = {}

for apt in apartment:
    apartment_city_map[apt[1]] = apt[3]


def generate_pesel():
    # Generate a random birth date
    year = random.randint(1940, 2006)
    month = random.randint(1, 12)
    day = random.randint(1, 28)

    # Format date as YYMMDD
    birth_date = f"{year % 100:02d}{month:02d}{day:02d}"

    # Generate the serial number (4 digits), odd for male, even for female
    gender_serial = random.randint(0, 9999)  # Random 4 digits

    # Gender control (even for female, odd for male)
    gender_digit = random.randint(0, 9)
    if gender_digit % 2 == 0:
        gender_serial += 1  # Ensure even for female

    # Add gender to serial number (4 digits)
    serial = f"{gender_serial:04d}"

    # Combine birth date and serial number
    pesel_base = birth_date + serial

    # Calculate checksum (11th digit)
    weights = [1, 3, 7, 9, 1, 3, 7, 9, 1, 3]
    total = sum(int(pesel_base[i]) * weights[i] for i in range(10))

    # Calculate the 11th digit (checksum)
    checksum = (10 - (total % 10)) % 10

    # Return full PESEL number
    pesel = pesel_base + str(checksum)
    return pesel

Gender_data = ["Male", "Female", "Other"]

start_date = datetime.strptime("2020-01-01", "%Y-%m-%d")
end_date = datetime.strptime("2025-01-01", "%Y-%m-%d")

#Owner table

owner = []
owner_pesels = []
number_of_rows_1o = 90

for i in range(1, number_of_rows_1o + 1):
  pesel = generate_pesel()
  surname = fake.last_name()
  gender = random.choice(Gender_data)
  if gender == "Male":
    name = fake.first_name_male()
  elif gender == "Female":
    name = fake.first_name_female()
  else:
    name = fake.first_name_nonbinary()
  phone = f"+48 {random.randint(100, 999)} {random.randint(100, 999)} {random.randint(100, 999)}"
  date = fake.date_between(start_date=start_date, end_date=end_date)

  owner.append([pesel, name, surname, gender, phone, date])
  owner_pesels.append(pesel)

#Real Estate Agent table

agent = []
agent_pesels = []
agents_by_city = {}

for i in range(1, number_of_rows_1 + 1):
  pesel = generate_pesel()
  surname = fake.last_name()
  gender = random.choice(Gender_data)
  if gender == "Male":
    name = fake.first_name_male()
  elif gender == "Female":
    name = fake.first_name_female()
  else:
    name = fake.first_name_nonbinary()
  phone = f"+48 {random.randint(100, 999)} {random.randint(100, 999)} {random.randint(100, 999)}"
  date = fake.date_between(start_date=start_date, end_date=end_date)
  city = list(city_data.keys())[i % 5]
  email = fake.email()

  agent.append([pesel, name, surname, gender, date, city, email, phone])
  agent_pesels.append(pesel)
  agents_by_city.setdefault(city, []).append(pesel)


#Selling process table

selling_process = []
before_deal = 0.03
No_deal_prob = 0.1
Sale_no = 0.20
cancellation_prob = 0.05
buyer_number = 0

for i in range(1, number_of_rows_1 + 1):
  id = i

  start = fake.date_between(start_date=start_date, end_date=end_date)

  if random.random() < before_deal:
    agreement = None
  elif random.random() > No_deal_prob:
    agreement = "Deal"
  else:
    agreement = "No Deal"

  if agreement == "No Deal" or agreement is None or random.random()<Sale_no:
    sale_date = None
  else:
    sale_date = fake.date_between(start_date=start, end_date=start + timedelta(days=30))

  if agreement == "No Deal":
    end = fake.date_between(start_date=start, end_date=start + timedelta(days=30))
  elif random.random() < cancellation_prob and sale_date == None:
    end = fake.date_between(start_date=start, end_date=start + timedelta(days=30))
  elif sale_date is None or random.random() < Sale_no:
    end = None
  else:
    end = fake.date_between(start_date=sale_date + timedelta(days=20), end_date=sale_date + timedelta(days=150))

  if agreement == "No Deal" or (end is not None and random.random() < 2*cancellation_prob):
    status = "Cancelled"
  elif agreement != "No Deal" and sale_date is None:
    status = "Pre-Sale"
  elif agreement == "Deal" and end is None:
    status = random.choice(["For Sale", "Under Offer"])
  elif agreement == "Deal" and end is not None:
    status = "Sold"

  if agreement != "Deal":
    listing_price = None
    final_price = None
    commision_rate = None
  elif status != "Sold":
    listing_price = round(random.uniform(500.0, 15000.0), 3)
    final_price = None
    commision_rate = round(random.uniform(2.0, 6.0), 1)
  else:
    listing_price = round(random.uniform(500.0, 15000.0), 3)
    final_price = round(random.uniform(0.9 * listing_price, 1.1 * listing_price), 3)
    commision_rate = round(random.uniform(2.0, 6.0), 1)

  if random.random() < 0.3:
    condition = random.choice(["very poor", "poor", "very good"])
  else:
    condition = random.choice(["acceptable", "good"])

  if i < number_of_rows_1 - 20:
    apartment_address = apartment_addresses[i]
  else:
    apartment_address = random.choice(apartment_addresses)
  apartment_city = apartment_city_map.get(apartment_address)
  if i < number_of_rows_1o:
    owner_pesel = owner_pesels[i-1]
  else:
    owner_pesel = random.choice(owner_pesels)
  agent_pesel = random.choice(agents_by_city[apartment_city])
  if status != "Sold" and status != "Under Offer":
    buyer_pesel = None
  else:
    buyer_pesel = 0
    buyer_number += 1

  selling_process.append([id, start, sale_date, end, listing_price, final_price, commision_rate, condition, status, agreement, apartment_address, owner_pesel, buyer_pesel, agent_pesel])

#Buyer table
buyer = []
buyer_pesels = []

for i in range(1, (buyer_number + 1) - 5):
  pesel = generate_pesel()
  surname = fake.last_name()
  gender = random.choice(Gender_data)
  if gender == "Male":
    name = fake.first_name_male()
  elif gender == "Female":
    name = fake.first_name_female()
  else:
    name = fake.first_name_nonbinary()
  phone = f"+48 {random.randint(100, 999)} {random.randint(100, 999)} {random.randint(100, 999)}"
  date = fake.date_between(start_date=start_date, end_date=end_date)

  buyer.append([pesel, name, surname, gender, phone, date])
  buyer_pesels.append(pesel)

number_of_used_buyers = 0

for i in range(1, number_of_rows_1 + 1):
  if selling_process[i-1][12] is not None:
    if number_of_used_buyers < len(buyer_pesels):
      selling_process[i-1][12] = buyer_pesels[number_of_used_buyers]
      number_of_used_buyers +=1
    else:
      selling_process[i-1][12] = random.choice(buyer_pesels)

pd.DataFrame(apartment, columns=["ID", "Address", "District", "City", "Postal_Code", "Size", "Floor_Number", "Number_Of_Rooms", "Year_Of_Construction"]).to_csv("1apartment.csv", index=False, header=False, sep="|", lineterminator=";")
pd.DataFrame(owner, columns=["PESEL", "Name", "Surname", "Gender", "Phone_Number", "First_Contact_Date"]).to_csv("1owner.csv", index=False, header=False, sep="|", lineterminator=";")
pd.DataFrame(buyer, columns=["PESEL", "Name", "Surname", "Gender", "Phone_Number", "First_Contact_Date"]).to_csv("1buyer.csv", index=False, header=False, sep="|", lineterminator=";")
pd.DataFrame(agent, columns=["PESEL", "Name", "Surname", "Gender", "Date_Of_Joining", "City_Of_Agency", "Email", "Phone_Number"]).to_csv("1agent.csv", index=False, header=False, sep="|", lineterminator=";")
pd.DataFrame(selling_process, columns=["ID", "Start_Date", "For_Sale_Start_Date", "End_Date", "Listing_Price", "Final_Price", "Commission_Rate", "Condition_Of_Apartment", "Status", "Agreement", "Apartment_Address", "Owner_PESEL", "Buyer_PESEL", "Real_Estate_Agent_PESEL"]).to_csv("1selling_process.csv", index=False, header=False, sep="|", lineterminator=";")

# Create Sheet 1: Agency Branch Offices
branch_offices = []
office_id = 1

for city in city_data:
    district, postal_code = random.choice(city_data[city])
    street_name = random.choice(city_district_streets[city][district])
    address = street_name + " " + str(random.randint(1, 100))
    # Count occurrences of the city in the 'City_Of_Agency' column of the 'agent' list
    num_agents = len([a for a in agent if a[5] == city])
    num_apartments = len([a for a in apartment if a[3] == city])

    branch_offices.append([office_id, city, address, postal_code, num_agents, num_apartments])
    office_id += 1

# Create Sheet 2: Interested Buyers per Apartment
interested_buyers = []
status_options = ["viewing", "negotiating price", "contract signing", "payment", "withdrawn"]

# Get a list of apartments from the 'selling_process' list
apartments_from_database = [item[10] for item in selling_process]  # apartment address

for apartment_address in apartments_from_database:
    # Get owner_pesel and start_date from selling_process
    owner_pesel = next((item[11] for item in selling_process if item[10] == apartment_address), None)
    start_date = next((item[1] for item in selling_process if item[10] == apartment_address), None)
    # Get the selling process status for this apartment
    selling_process_status = next((item[8] for item in selling_process if item[10] == apartment_address), None)
    # Get the buyer associated with this apartment from 'selling_process' list
    buyer_from_database = next((item[12] for item in selling_process if item[10] == apartment_address), None)
    # Get the city from the apartment list:
    city = next((item[3] for item in apartment if item[1] == apartment_address), None)

    first_buyer_status = None  # Track status of the first buyer

    if buyer_from_database:
        # buyer_from_database is a PESEL, get the buyer details from the 'buyer' list
        buyer_details = next((b for b in buyer if b[0] == buyer_from_database), None)

        if buyer_details:
            # Assign status based on selling process status
            if selling_process_status == "Sold":
                status = "payment"
            elif selling_process_status == "Under Offer":
                status = "contract signing"
            elif selling_process_status == "For Sale": # no buyer
                status = random.choice(["viewing", "negotiating price"])
            elif selling_process_status == "Cancelled": # no
                status = random.choice([s for s in status_options if s not in ["payment"]])
            else: # pre-sale - apartment not stored in selling process
                status = "viewing"

            # Store first buyer's status
            first_buyer_status = status

            interested_buyers.append([
                buyer_details[1],  # Name
                buyer_details[2],  # Surname
                buyer_details[4],  # Phone number
                city,
                apartment_address,  # Apartment address
                owner_pesel,
                start_date,
                first_buyer_status,
                buyer_details[5]  # First contact date
            ])

    # Add two randomly generated buyers for each apartment
    for _ in range(2):
        pesel = generate_pesel()
        surname = fake.last_name()
        gender = random.choice(Gender_data)
        name = (
            fake.first_name_male() if gender == "Male"
            else fake.first_name_female() if gender == "Female"
            else fake.first_name_nonbinary()
        )
        phone = f"+48 {random.randint(100, 999)} {random.randint(100, 999)} {random.randint(100, 999)}"
        date = fake.date_between(start_date=start_date, end_date=end_date)

        # Determine status options based on first buyer's status
        if first_buyer_status in ["contract signing", "payment"]:
            status = "withdrawn"
        else:
            status = random.choice(["viewing", "withdrawn"])

        interested_buyers.append([
            name,  # Name
            surname,  # Surname
            phone,  # Phone number
            city,
            apartment_address,  # Apartment address
            owner_pesel,
            start_date,
            status,
            date  # First contact date
        ])



# Create DataFrames
df_branches = pd.DataFrame(branch_offices, columns=[
    "Office ID",
    "City",
    "Address",
    "Postal Code",
    "Number of Agents",
    "Number of Apartments"
])

df_buyers = pd.DataFrame(interested_buyers, columns=[
    "Name",
    "Surname",
    "Phone Number",
    "City",
    "Interested Apartment",
    "Owner PESEL",
    "Start Date of the Selling Process",
    "Status",
    "Buyer First Contact Date"
])

# Create Excel writer and save to file
with pd.ExcelWriter("1.agency_data.xlsx") as writer:
    df_branches.to_excel(writer, sheet_name="Branch Offices", index=False)
    df_buyers.to_excel(writer, sheet_name="Interested Buyers", index=False)

print("Excel file '1.agency_data.xlsx' has been created with 2 sheets.")

# Snapshot 2

#apartment 2
#updating numbr of rooms in some apartments for example because of renovations
apartment[4][7] += 1
apartment[16][7] -= 1
apartment[20][7] += 1
apartment[55][7] -= 1
apartment[71][7] += 1

number_of_rows_2 = 20

for i in range(number_of_rows_1 + 1, number_of_rows_1 + number_of_rows_2 + 1):
  apartment_id = i
  city = random.choice(list(city_data.keys()))
  district, postal_code = random.choice(city_data[city])
  street_name = random.choice(city_district_streets[city][district])
  address = street_name + " " + str(random.randint(1, 100)) + "/" + str(random.randint(1,100))
  size = round(random.uniform(20.0, 150.0), 2)
  floor_number = random.randint(0, 30)
  if size < 40:
      number_of_rooms = random.choice([1, 2])
  elif size < 80:
      number_of_rooms = random.choice([2, 3])
  else:
      number_of_rooms = random.choice([3, 4, 5])
  start_date = datetime.strptime("1950-01-01", "%Y-%m-%d")
  end_date = datetime.strptime("2025-01-01", "%Y-%m-%d")
  year = fake.date_between(start_date=start_date, end_date=end_date)

  apartment.append([apartment_id, address, district, city, postal_code, size, floor_number, number_of_rooms, year])
  apartment_addresses.append(address)

for apt in apartment:
    apartment_city_map[apt[1]] = apt[3]

#owner 2

#updating phone numbers of some owners
owner[3][4] = f"+48 {random.randint(100, 999)} {random.randint(100, 999)} {random.randint(100, 999)}"
owner[15][4] = f"+48 {random.randint(100, 999)} {random.randint(100, 999)} {random.randint(100, 999)}"

for i in range(1, number_of_rows_2 + 1):
  pesel = generate_pesel()
  surname = fake.last_name()
  gender = random.choice(Gender_data)
  if gender == "Male":
    name = fake.first_name_male()
  elif gender == "Female":
    name = fake.first_name_female()
  else:
    name = fake.first_name_nonbinary()
  phone = f"+48 {random.randint(100, 999)} {random.randint(100, 999)} {random.randint(100, 999)}"
  date = fake.date_between(start_date=start_date, end_date=end_date)
  owner.append([pesel, name, surname, gender, phone, date])
  owner_pesels.append(pesel)

#Real estate agent 2

#updating the city of agency for some real estate agents
for i in range(1, number_of_rows_1 + 1, int(number_of_rows_1/5)):
  current_city = agent[i][5]  # Get current city
  current_index = list(city_data.keys()).index(current_city)
  new_index = (current_index + 1) % len(list(city_data.keys()))
  agent[i][5] = list(city_data.keys())[new_index]

for i in range(1, number_of_rows_2 + 1):
  pesel = generate_pesel()
  surname = fake.last_name()
  gender = random.choice(Gender_data)
  if gender == "Male":
    name = fake.first_name_male()
  elif gender == "Female":
    name = fake.first_name_female()
  else:
    name = fake.first_name_nonbinary()
  phone = f"+48 {random.randint(100, 999)} {random.randint(100, 999)} {random.randint(100, 999)}"
  date = fake.date_between(start_date=start_date, end_date=end_date)
  city = list(city_data.keys())[i % 5]
  email = fake.email()

  agent.append([pesel, name, surname, gender, date, city, email, phone])
  agent_pesels.append(pesel)
  agents_by_city.setdefault(city, []).append(pesel)

#Selling process 2

#updating the statuses of some selling processes and with that dates and final prices
for i in range(0, number_of_rows_1 + 1):
  if (selling_process[i][8] == "For Sale" or selling_process[i][8] == "Under Offer") and (i % 2 == 0 or i < 10):
    selling_process[i][8] = "Sold"
    selling_process[i][3] = fake.date_between(start_date=selling_process[i][2] + timedelta(days=20), end_date=selling_process[i][2] + timedelta(days=150))
    selling_process[i][5] = round(random.uniform(0.9 * selling_process[i][4], 1.1 * selling_process[i][4]), 3)
    selling_process[i][12] = random.choice(buyer_pesels)
    i += 5
  elif selling_process[i][8] == "Pre-Sale" and i % 3 == 0:
    selling_process[i][8] = "For Sale"
    selling_process[i][2] = fake.date_between(start_date=selling_process[i][1] + timedelta(days=5), end_date=selling_process[i][1] + timedelta(days=30))
    i += 10
  elif selling_process[i][8] != "Sold" and selling_process[i][8] != "Cancelled" and i % 4 == 0:
    if selling_process[i][8] == "Pre-Sale":
      selling_process[i][3] = fake.date_between(start_date=selling_process[i][1] + timedelta(days=20), end_date=selling_process[i][1] + timedelta(days=150))
    else:
      selling_process[i][3] = fake.date_between(start_date=selling_process[i][2] + timedelta(days=20), end_date=selling_process[i][2] + timedelta(days=150))
    selling_process[i][8] = "Cancelled"
    i += 10
  if i >= number_of_rows_1 - 1:
    break

iteration = 0
buyer_number2 = 0

for i in range(number_of_rows_1 + 1, number_of_rows_1 + number_of_rows_2 + 1):
  id = i

  start = fake.date_between(start_date=start_date, end_date=end_date)

  if random.random() < before_deal:
    agreement = None
  elif random.random() > No_deal_prob:
    agreement = "Deal"
  else:
    agreement = "No Deal"

  if agreement == "No Deal" or agreement is None or random.random()<Sale_no:
    sale_date = None
  else:
    sale_date = fake.date_between(start_date=start, end_date=start + timedelta(days=30))

  if agreement == "No Deal":
    end = fake.date_between(start_date=start, end_date=start + timedelta(days=30))
  elif random.random() < cancellation_prob and sale_date == None:
    end = fake.date_between(start_date=start, end_date=start + timedelta(days=30))
  elif sale_date is None or random.random() < Sale_no:
    end = None
  else:
    end = fake.date_between(start_date=sale_date + timedelta(days=20), end_date=sale_date + timedelta(days=150))

  if agreement == "No Deal" or (end is not None and random.random() < 2*cancellation_prob):
    status = "Cancelled"
  elif agreement != "No Deal" and sale_date is None:
    status = "Pre-Sale"
  elif agreement == "Deal" and end is None:
    status = random.choice(["For Sale", "Under Offer"])
  elif agreement == "Deal" and end is not None:
    status = "Sold"

  if agreement != "Deal":
    listing_price = None
    final_price = None
    commision_rate = None
  elif status != "Sold":
    listing_price = round(random.uniform(500.0, 15000.0), 3)
    final_price = None
    commision_rate = round(random.uniform(2.0, 6.0), 1)
  else:
    listing_price = round(random.uniform(500.0, 15000.0), 3)
    final_price = round(random.uniform(0.9 * listing_price, 1.1 * listing_price), 3)
    commision_rate = round(random.uniform(2.0, 6.0), 1)

  if random.random() < 0.3:
    condition = random.choice(["very poor", "poor", "very good"])
  else:
    condition = random.choice(["acceptable", "good"])

  apartment_address = apartment_addresses[i - 1]
  apartment_city = apartment_city_map.get(apartment_address)
  owner_pesel = owner_pesels[(number_of_rows_1o) + iteration]
  agent_pesel = random.choice(agents_by_city[apartment_city])
  if status != "Sold" and status != "Under Offer":
    buyer_pesel = None
  else:
    buyer_pesel = 0
    buyer_number2 += 1

    iteration += 1

  selling_process.append([id, start, sale_date, end, listing_price, final_price, commision_rate, condition, status, agreement, apartment_address, owner_pesel, buyer_pesel, agent_pesel])

#Buyer table 2

buyer[2][4] = f"+48 {random.randint(100, 999)} {random.randint(100, 999)} {random.randint(100, 999)}"
buyer[11][4] = f"+48 {random.randint(100, 999)} {random.randint(100, 999)} {random.randint(100, 999)}"

for i in range(1, buyer_number2 + 1):
  pesel = generate_pesel()
  surname = fake.last_name()
  gender = random.choice(Gender_data)
  if gender == "Male":
    name = fake.first_name_male()
  elif gender == "Female":
    name = fake.first_name_female()
  else:
    name = fake.first_name_nonbinary()
  phone = f"+48 {random.randint(100, 999)} {random.randint(100, 999)} {random.randint(100, 999)}"
  date = fake.date_between(start_date=start_date, end_date=end_date)

  buyer.append([pesel, name, surname, gender, phone, date])
  buyer_pesels.append(pesel)

for i in range(number_of_rows_1 + 1, number_of_rows_1 + number_of_rows_2 + 1):
  if selling_process[i-1][12] is not None:
    selling_process[i-1][12] = buyer_pesels[buyer_number]

pd.DataFrame(apartment, columns=["ID", "Address", "District", "City", "Postal_Code", "Size", "Floor_Number", "Number_Of_Rooms", "Year_Of_Construction"]).to_csv("2apartment.csv", index=False, header=False, sep="|", lineterminator=";")
pd.DataFrame(owner, columns=["PESEL", "Name", "Surname", "Gender", "Phone_Number", "First_Contact_Date"]).to_csv("2owner.csv", index=False, header=False, sep="|", lineterminator=";")
pd.DataFrame(buyer, columns=["PESEL", "Name", "Surname", "Gender", "Phone_Number", "First_Contact_Date"]).to_csv("2buyer.csv", index=False, header=False, sep="|", lineterminator=";")
pd.DataFrame(agent, columns=["PESEL", "Name", "Surname", "Gender", "Date_Of_Joining", "City_Of_Agency", "Email", "Phone_Number"]).to_csv("2agent.csv", index=False, header=False, sep="|", lineterminator=";")
pd.DataFrame(selling_process, columns=["ID", "Start_Date", "For_Sale_Start_Date", "End_Date", "Listing_Price", "Final_Price", "Commission_Rate", "Condition_Of_Apartment", "Status", "Agreement", "Apartment_Address", "Owner_PESEL", "Buyer_PESEL", "Real_Estate_Agent_PESEL"]).to_csv("2selling_process.csv", index=False, header=False, sep="|", lineterminator=";")


# EXCEL snapshot 2 - Modified version that works with the existing interested_buyers list

# Sheet 1: Agency Branch Offices
# Create a copy of the original branch_offices list
branch_offices_snapshot2 = [row.copy() for row in branch_offices]

# Choose a random branch office to update
random_branch_index = random.randint(0, len(branch_offices_snapshot2) - 1)

# Change the address of the chosen branch office
city = branch_offices_snapshot2[random_branch_index][1]  # City is at index 1
district, postal_code = random.choice(city_data[city])
street_name = random.choice(city_district_streets[city][district])
new_address = street_name + " " + str(random.randint(1, 100))
branch_offices_snapshot2[random_branch_index][2] = new_address  # Address is at index 2

# Update the number of agents and apartments for all branch offices
for office in branch_offices_snapshot2:
    city = office[1]
    # Count agents and apartments for this city
    num_agents = len([a for a in agent if a[5] == city])
    num_apartments = len([a for a in apartment if a[3] == city])
    office[4] = num_agents  # Number of agents is at index 4
    office[5] = num_apartments  # Number of apartments is at index 5

# Sheet 2: Interested Buyers per Apartment
# Create a copy of the original interested_buyers list
interested_buyers_snapshot2 = [row.copy() for row in interested_buyers]

# Get a list of unique apartment addresses from the interested buyers list
list_apartments = list(set([buyer[4] for buyer in interested_buyers_snapshot2]))

# Choose 10 random apartments
random_apartments = random.sample(list_apartments, min(10, len(list_apartments)))

# Add new buyers for these apartments
for apartment_address in random_apartments:
    # Find the existing buyer info for this apartment to get city, owner_pesel, etc.
    existing_buyers = [b for b in interested_buyers_snapshot2 if b[4] == apartment_address]
    if existing_buyers:
        first_buyer = existing_buyers[0]
        city = first_buyer[3]
        owner_pesel = first_buyer[5]
        start_date = first_buyer[6]

        # Generate new buyer data
        gender = random.choice(Gender_data)
        if gender == "Male":
            name = fake.first_name_male()
        elif gender == "Female":
            name = fake.first_name_female()
        else:
            name = fake.first_name_nonbinary()

        surname = fake.last_name()
        phone = f"+48 {random.randint(100, 999)} {random.randint(100, 999)} {random.randint(100, 999)}"
        date = fake.date_between(start_date=start_date, end_date=end_date)

        # Determine status - if any existing buyer is in contract signing or payment, new buyer is withdrawn
        existing_statuses = [b[7] for b in existing_buyers]
        if "contract signing" in existing_statuses or "payment" in existing_statuses:
            status = "withdrawn"
        else:
            status = random.choice(["viewing", "withdrawn"])

        # Add the new buyer to the list
        interested_buyers_snapshot2.append([
            name,
            surname,
            phone,
            city,
            apartment_address,
            owner_pesel,
            start_date,
            status,
            date
        ])

# Change the surname of 5 random buyers
random_buyers_indices = random.sample(range(len(interested_buyers_snapshot2)), min(5, len(interested_buyers_snapshot2)))
for index in random_buyers_indices:
    interested_buyers_snapshot2[index][1] = fake.last_name()  # Surname is at index 1

# Create DataFrames for the snapshot
df_branches_snapshot2 = pd.DataFrame(branch_offices_snapshot2, columns=[
    "Office ID",
    "City",
    "Address",
    "Postal Code",
    "Number of Agents",
    "Number of Apartments"
])

df_buyers_snapshot2 = pd.DataFrame(interested_buyers_snapshot2, columns=[
    "Name",
    "Surname",
    "Phone Number",
    "City",
    "Interested Apartment",
    "Owner PESEL",
    "Start Date of the Selling Process",
    "Status",
    "Buyer First Contact Date"
])

# Save the snapshot to a new Excel file
with pd.ExcelWriter("2.agency_data.xlsx") as writer:
    df_branches_snapshot2.to_excel(writer, sheet_name="Branch Offices", index=False)
    df_buyers_snapshot2.to_excel(writer, sheet_name="Interested Buyers", index=False)

print("Excel file '2.agency_data.xlsx' has been created with updated data.")
```
